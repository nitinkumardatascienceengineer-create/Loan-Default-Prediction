# Logistic Regression 

```python
import numpy as np
import pandas as pd

df = pd.read_csv('loan_default_dataset.csv')

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
confusion_matrix,
classification_report,
)

df.head()

df = df.drop_duplicates()

df.columns

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'median'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
        ("scaler",StandardScaler())
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",LogisticRegression(max_iter = 1000, random_state=42,class_weight='balanced'))
])

pipeline.fit(X_train,y_train)


y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:,1]

acc = accuracy_score(y_test,y_pred)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob)
print(f"roc_auc Score = {roc_auc : .4f}")

cr = classification_report(y_test,y_pred)
print(f"Classification Report = {cr}")

cm = confusion_matrix(y_test,y_pred)

print(cm)
```

# Decision tree

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
confusion_matrix,
classification_report,
)

df = pd.read_csv("loan_default_dataset.csv")

df.shape

df = df.drop_duplicates()

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",DecisionTreeClassifier(random_state = 42,
                                    max_depth = 7,
                                    criterion = 'gini',
                                    max_leaf_nodes = 10,
                                    min_samples_split = 20,class_weight='balanced'))
])

pipeline.fit(X_train,y_train)


y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:,1]

print('Base Decision Tree')
acc = accuracy_score(y_test,y_pred)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob)
print(f"roc_auc Score = {roc_auc : .4f}")

cr = classification_report(y_test,y_pred)
print(cr)

cm = confusion_matrix(y_test,y_pred)

print(cm)

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",DecisionTreeClassifier(random_state = 42,class_weight='balanced'))
])


param_grid = {
    'model__max_depth' : [7,10,15,17],
'model__max_leaf_nodes': [10, 20, 50, None],
'model__min_samples_leaf': [1, 2, 4, 8],
'model__min_samples_split' : [5,15,30],
'model__criterion' : ['gini','entropy','log_loss']
}

grid = GridSearchCV(
    estimator = pipeline,
    param_grid = param_grid,
    cv = 5,
    scoring = "roc_auc",
    n_jobs = -1 
)
grid.fit(X_train,y_train)


y_bass_pred = grid.predict(X_test)
y_bass_prob = grid.predict_proba(X_test)[:,1]

print('gridsearch decision tree model')

bass_acc = accuracy_score(y_test,y_bass_pred)
print(f"Accuracy Score = {bass_acc : .2f}")

bass_roc_auc = roc_auc_score(y_test,y_bass_prob)
print(f"roc_auc Score = {bass_roc_auc : .4f}")

bass_cr = classification_report(y_test,y_bass_pred)
print(bass_cr)

bass_cm = confusion_matrix(y_test,y_pred)

print(bass_cm)
```

# Random Forest

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
confusion_matrix,
classification_report,
)

df = pd.read_csv("loan_default_dataset.csv")

df.shape

df = df.drop_duplicates()

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",RandomForestClassifier(
        n_estimators= 1000,
        max_depth = 5,
        min_samples_split=10,
        min_samples_leaf = 2,
        max_features = "sqrt",
        random_state = 42,
        class_weight='balanced')
    )
])

pipeline.fit(X_train,y_train)


y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:,1]

print('Base Random forest')
acc = accuracy_score(y_test,y_pred)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob)
print(f"roc_auc Score = {roc_auc : .4f}")

cr = classification_report(y_test,y_pred)
print(cr)

cm = confusion_matrix(y_test,y_pred)

print(cm)

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",RandomForestClassifier(random_state = 42,class_weight='balanced'))
])


param_grid = {
    "model__criterion" : [
        "gini",
        "entropy"
    ],
    "model__n_estimators" : [100,200,500,700,1000],
    "model__max_depth" : [2,3,4,5,6,8],
    "model__min_samples_split" : [2,5,7,10],
    "model__min_samples_leaf" : [1,2,5,7]
}

grid = GridSearchCV(
    estimator = pipeline,
    param_grid = param_grid,
    cv = 5,
    scoring = "roc_auc",
    n_jobs = -1 
)
grid.fit(X_train,y_train)


y_bass_pred = grid.predict(X_test)
y_bass_prob = grid.predict_proba(X_test)[:,1]

print('gridsearch random forest model')

bass_acc = accuracy_score(y_test,y_bass_pred)
print(f"Accuracy Score = {bass_acc : .2f}")

bass_roc_auc = roc_auc_score(y_test,y_bass_prob)
print(f"roc_auc Score = {bass_roc_auc : .4f}")

bass_cr = classification_report(y_test,y_bass_pred)
print(bass_cr)

bass_cm = confusion_matrix(y_test,y_pred)

print(bass_cm)
```
