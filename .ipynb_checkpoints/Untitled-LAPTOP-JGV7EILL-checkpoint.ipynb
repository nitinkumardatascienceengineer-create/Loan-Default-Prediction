# Logistic Regression 

```python
import numpy as np
import pandas as pd

df = pd.read_csv('loan_default_dataset.csv')

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
confusion_matrix,
classification_report,
)

df.head()

df = df.drop_duplicates()

df.columns

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'median'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
        ("scaler",StandardScaler())
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",LogisticRegression(max_iter = 1000, random_state=42,class_weight='balanced'))
])

pipeline.fit(X_train,y_train)


y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:,1]

acc = accuracy_score(y_test,y_pred)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob)
print(f"roc_auc Score = {roc_auc : .4f}")

cr = classification_report(y_test,y_pred)
print(f"Classification Report = {cr}")

cm = confusion_matrix(y_test,y_pred)

print(cm)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
confusion_matrix,
classification_report,
)

df = pd.read_csv("loan_default_dataset.csv")

df.shape

df = df.drop_duplicates()

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",DecisionTreeClassifier(random_state = 42,
                                    max_depth = 7,
                                    criterion = 'gini',
                                    max_leaf_nodes = 10,
                                    min_samples_split = 20,class_weight='balanced'))
])

pipeline.fit(X_train,y_train)


y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:,1]

print('Base Decision Tree')
acc = accuracy_score(y_test,y_pred)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob)
print(f"roc_auc Score = {roc_auc : .4f}")

cr = classification_report(y_test,y_pred)
print(cr)

cm = confusion_matrix(y_test,y_pred)

print(cm)

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",DecisionTreeClassifier(random_state = 42,class_weight='balanced'))
])


param_grid = {
    'model__max_depth' : [7,10,15,17],
'model__max_leaf_nodes': [10, 20, 50, None],
'model__min_samples_leaf': [1, 2, 4, 8],
'model__min_samples_split' : [5,15,30],
'model__criterion' : ['gini','entropy','log_loss']
}

grid = GridSearchCV(
    estimator = pipeline,
    param_grid = param_grid,
    cv = 5,
    scoring = "roc_auc",
    n_jobs = -1 
)
grid.fit(X_train,y_train)


y_bass_pred = grid.predict(X_test)
y_bass_prob = grid.predict_proba(X_test)[:,1]

print('gridsearch decision tree model')

bass_acc = accuracy_score(y_test,y_bass_pred)
print(f"Accuracy Score = {bass_acc : .2f}")

bass_roc_auc = roc_auc_score(y_test,y_bass_prob)
print(f"roc_auc Score = {bass_roc_auc : .4f}")

bass_cr = classification_report(y_test,y_bass_pred)
print(bass_cr)

bass_cm = confusion_matrix(y_test,y_pred)

print(bass_cm)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
confusion_matrix,
classification_report,
)

df = pd.read_csv("loan_default_dataset.csv")

df.shape

df = df.drop_duplicates()

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",RandomForestClassifier(
        n_estimators= 1000,
        max_depth = 5,
        min_samples_split=10,
        min_samples_leaf = 2,
        max_features = "sqrt",
        random_state = 42,
        class_weight='balanced')
    )
])

pipeline.fit(X_train,y_train)


y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:,1]

print('Base Random forest')
acc = accuracy_score(y_test,y_pred)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob)
print(f"roc_auc Score = {roc_auc : .4f}")

cr = classification_report(y_test,y_pred)
print(cr)

cm = confusion_matrix(y_test,y_pred)

print(cm)

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("encoder",OneHotEncoder(handle_unknown = 'ignore')),
    ("Impute",SimpleImputer(strategy = 'most_frequent'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",RandomForestClassifier(random_state = 42,class_weight='balanced'))
])


param_grid = {
    "model__criterion" : [
        "gini",
        "entropy"
    ],
    "model__n_estimators" : [100,200,500,1000],
    "model__max_depth" : [2,3,4,8],
    "model__min_samples_split" : [2,5,7,10],
    "model__min_samples_leaf" : [1,2,5,7]
}

grid = GridSearchCV(
    estimator = pipeline,
    param_grid = param_grid,
    cv = 5,
    scoring = "roc_auc",
    n_jobs = -1 
)
grid.fit(X_train,y_train)


y_bass_pred = grid.predict(X_test)
y_bass_prob = grid.predict_proba(X_test)[:,1]

print('gridsearch random forest model')

bass_acc = accuracy_score(y_test,y_bass_pred)
print(f"Accuracy Score = {bass_acc : .2f}")

bass_roc_auc = roc_auc_score(y_test,y_bass_prob)
print(f"roc_auc Score = {bass_roc_auc : .4f}")

bass_cr = classification_report(y_test,y_bass_pred)
print(bass_cr)

bass_cm = confusion_matrix(y_test,y_pred)

print(bass_cm)
```

# LightGBM

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
import lightgbm as lgb
from sklearn.pipeline import Pipeline
from sklearn.metrics import(
accuracy_score,
roc_auc_score,
precision_score,
recall_score,
f1_score,
classification_report,
confusion_matrix
)

df = pd.read_csv("loan_default_dataset.csv")

df.shape

df = df.drop_duplicates()

X = df.drop(columns = ['Loan_Default','Loan_ID','Customer_ID'])
y = df['Loan_Default']

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify = y
)

categorical_cols = X.select_dtypes(include = 'object').columns

numerical_features = X.select_dtypes(include = 'number').columns

Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
            ("Impute",SimpleImputer(strategy = 'most_frequent')),
            ("encoder",OneHotEncoder(handle_unknown = 'ignore'))
    
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("model",lgb.LGBMClassifier(
        objective = "binary",
        n_estimators = 1000,
        learning_rate = 0.03,
        num_leaves = 31,
        max_depth = -1,
        min_child_samples = 30,
        subsample = 0.8,
        subsample_freq=1,
        colsample_bytree = 0.8,
        reg_alpha = 0.1,
        reg_lambda = 1.0,
        random_state = 42,
        class_weight="balanced",
        n_jobs = -1,
        verbosity = -1,
    )
    )
])

# Split training data into training and validation sets
X_train_new, X_val, y_train_new, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Preprocess the data
model = pipeline.named_steps['model']
processor = pipeline.named_steps['processor']

X_train_processed = processor.fit_transform(X_train_new)
X_val_processed = processor.transform(X_val)
X_test_processed = processor.transform(X_test)

eval_result = {}

# Train with early stopping
model.fit(
    X_train_processed,
    y_train_new,
    eval_set=[(X_val_processed, y_val)],
    eval_metric='auc',
    callbacks=[
        lgb.early_stopping(50),
        lgb.record_evaluation(eval_result),
        lgb.log_evaluation(50)
    ]
)

# Plot validation AUC
plt.plot(eval_result['valid_0']['auc'])
plt.xlabel('Boosting Round')
plt.ylabel('Validation AUC')
plt.title('LightGBM Learning Curve')
plt.show()

print("Best iteration:", model.best_iteration_)


y_pred = model.predict(
    X_test_processed,
    num_iteration=model.best_iteration_
)

y_prob = model.predict_proba(
    X_test_processed,
    num_iteration=model.best_iteration_
)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred))

print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test,y_pred)
sns.heatmap(
    cm,
    annot = True,
    fmt = "d"
)
plt.title("LightGBM Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

from sklearn.metrics import roc_curve, roc_auc_score

fpr,tpr,thresholds = roc_curve(
    y_test,y_prob
)
auc_score = roc_auc_score(y_test,y_prob)

plt.plot(fpr,tpr,
        label = f"LightGBM AUC ={auc_score : .3f}")
plt.plot(
    [0,1],
    [0,1],
    linestyle = "--"
)
plt.title("LightGBM ROC Curve")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.legend()
plt.show()

from sklearn.metrics import precision_recall_curve

recall,precision,thresholds = precision_recall_curve(
    y_test,y_prob
)
plt.plot(recall,precision)

plt.title("LightGBM Precision-Recall Curve")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.show()


feature_names = pipeline.named_steps['processor'].get_feature_names_out()

gain_importance = model.booster_.feature_importance(
    importance_type="gain"
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Gain": gain_importance
}).sort_values(
    by="Gain",
    ascending=False
)

print(feature_importance.head(15))

print(model.booster_.feature_importance(
    importance_type='split'
))

print(model.booster_.feature_importance(
    importance_type='gain'
))

lgb.plot_importance(
    model,
    importance_type = "gain",
    max_num_features = 15,
    figsize = (10,8)
)
plt.title("LightGBM Feature Importance - Gain")
plt.show()
```

```python
Preprocessor = ColumnTransformer(
    [
        ('cat',Pipeline([
        
    ("Impute",SimpleImputer(strategy = 'most_frequent')),
            ("encoder",OneHotEncoder(handle_unknown = 'ignore'))
        
    ]),categorical_cols),

     ('num',Pipeline([
    ("Impute",SimpleImputer(strategy = 'median')),
    ]),numerical_features)
        
    ]
)

pipeline = Pipeline([
    ("processor",Preprocessor),
    ("lgb_model",lgb.LGBMClassifier(
    objective = 'binary',
    random_state = 42,
    class_weight = 'balanced',
    n_jobs = -1
))
])


param_grid = {
    "lgb_model__n_estimators" : [200,400,600,1000],
    "lgb_model__learning_rate" : [0.01,0.03,0.05],
    "lgb_model__num_leaves" : [15,31,51],
    "lgb_model__max_depth" : [-1,5,8,12],
    "lgb_model__min_child_samples" : [20,30,50,70],
    "lgb_model__subsample" :[0.6,0.7,0.8,0.9],
    "lgb_model__colsample_bytree" : [0.6,0.7,0.8,0.9],
    "lgb_model__reg_alpha" : [0,0.1,0.5],
    "lgb_model__reg_lambda" : [0,0.1,1.0],
    "lgb_model__subsample_freq": [1]
}

from sklearn.model_selection import RandomizedSearchCV

random_search = RandomizedSearchCV(
    estimator = pipeline,
    param_distributions = param_grid,
    n_iter = 20,
    scoring = "roc_auc",
    cv = 3,
    verbose = 1,
    random_state = 42,
    n_jobs = 1
)
random_search.fit(X_train,
                  y_train,)

print(random_search.best_params_)
print(random_search.best_score_)

best_model = random_search.best_estimator_

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
import numpy as np
import pandas as pd

# Create validation data from the training set
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Refit the selected pipeline using only the fitting subset
best_model.fit(X_fit, y_fit)

# Predict default probabilities on validation data
y_prob_val = best_model.predict_proba(X_val)[:, 1]

results = []

for threshold in np.arange(0.10, 0.71, 0.01):
    y_pred = (y_prob_val >= threshold).astype(int)

    results.append({
        "Threshold": round(threshold, 2),
        "Precision": precision_score(
            y_val, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_val, y_pred, zero_division=0
        ),
        "F1": f1_score(
            y_val, y_pred, zero_division=0
        )
    })

threshold_results = pd.DataFrame(results)

# Choose the threshold with the highest F1-score
best_row = threshold_results.loc[
    threshold_results["F1"].idxmax()
]

best_threshold = best_row["Threshold"]

print("Best threshold:", best_threshold)
print(best_row)

print("\nTop 10 thresholds:")
print(
    threshold_results.sort_values(
        "F1", ascending=False
    ).head(10).round(3)
)

y_prob_grid = best_model.predict_proba(X_test)[:,1]
y_pred_grid = best_model.predict(X_test)

acc = accuracy_score(y_test,y_pred_grid)
print(f"Accuracy Score = {acc : .2f}")

roc_auc = roc_auc_score(y_test,y_prob_grid)
print(f"roc_auc Score = {roc_auc : .4f}")

print(f"precision = {precision_score(y_test,y_pred_grid)}")

print(f"recall = {recall_score(y_test,y_pred_grid)}")

print(f"f1 = {f1_score(y_test,y_pred_grid)}")

cr = classification_report(y_test,y_pred_grid,zero_division=0)
print(cr)

cm = confusion_matrix(y_test,y_pred_grid)
sns.heatmap(
    cm,
    annot = True,
    fmt = "d"
)
plt.title("LightGBM Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

from sklearn.metrics import roc_curve, roc_auc_score

fpr,tpr,thresholds = roc_curve(
    y_test,y_prob_grid
)
auc_score = roc_auc_score(y_test,y_prob_grid)

plt.plot(fpr,tpr,
        label = f"LightGBM AUC ={auc_score : .3f}")
plt.plot(
    [0,1],
    [0,1],
    linestyle = "--"
)
plt.title("LightGBM ROC Curve")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.legend()
plt.show()

from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_test, y_prob_grid
)

plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("LightGBM Precision-Recall Curve")
plt.show()

model = best_model.named_steps["lgb_model"]

processor = best_model.named_steps["processor"]

feature_names = processor.get_feature_names_out()

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Split_Importance": model.booster_.feature_importance(
        importance_type="split"
    ),
    "Gain_Importance": model.booster_.feature_importance(
        importance_type="gain"
    )
}).sort_values(
    by="Gain_Importance",
    ascending=False
)

print(feature_importance.head(15))

lgb.plot_importance(
    model,
    importance_type="gain",
    max_num_features=15,
    figsize=(10, 8)
)
plt.title("LightGBM Feature Importance - Gain")
plt.show()




y_pred_custom_grid = (
    y_prob_grid >= best_threshold
).astype(int)

print(classification_report(
    y_test,y_pred_custom_grid
))

from sklearn.model_selection import StratifiedKFold , cross_val_score

cv = StratifiedKFold(
    n_splits = 5,
    shuffle = True,
    random_state = 42
)
scores = cross_val_score(
    best_model,
    X_train,
    y_train,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1
)

print("CV Scores:", scores)
print("Mean AUC:", scores.mean())
print("STD:", scores.std())

import shap

model = best_model.named_steps["lgb_model"]
processor = best_model.named_steps["processor"]

X_test_processed = processor.transform(X_test)
feature_names = processor.get_feature_names_out()

# Convert sparse output to dense if necessary
if hasattr(X_test_processed, "toarray"):
    X_test_processed = X_test_processed.toarray()

X_test_shap = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test_shap)

# Handle binary-class output from different SHAP versions
if isinstance(shap_values, list):
    shap_values = shap_values[1]
elif isinstance(shap_values, np.ndarray) and shap_values.ndim == 3:
    shap_values = shap_values[:, :, 1]

shap.summary_plot(
    shap_values,
    X_test_shap,
    max_display=15
)
```

```python

import numpy as np
import pandas as pd
import lightgbm as lgb
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    RandomizedSearchCV,
    StratifiedKFold
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    precision_recall_curve
)

# 1. Separate fitting, validation and test data
# X_train and X_test should already come from your original
# stratified 80/20 train-test split.

X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# 2. Identify feature types using fitting data
categorical_cols = X_fit.select_dtypes(
    include=["object", "category", "bool"]
).columns

numerical_cols = X_fit.select_dtypes(
    include=["number"]
).columns

# 3. Build preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(
                    strategy="most_frequent"
                )),
                ("encoder", OneHotEncoder(
                    handle_unknown="ignore"
                ))
            ]),
            categorical_cols
        ),
        (
            "num",
            Pipeline([
                ("imputer", SimpleImputer(
                    strategy="median"
                ))
            ]),
            numerical_cols
        )
    ],
    remainder="drop"
)

# 4. Build LightGBM pipeline
pipeline = Pipeline([
    ("processor", preprocessor),
    ("lgb_model", lgb.LGBMClassifier(
        objective="binary",
        class_weight="balanced",
        random_state=42,
        n_jobs=1,
        verbosity=-1
    ))
])

# 5. Hyperparameter search space
param_distributions = {
    "lgb_model__n_estimators": [200, 400, 600, 800, 1000],
    "lgb_model__learning_rate": [0.01, 0.02, 0.03, 0.05, 0.1],
    "lgb_model__num_leaves": [15, 31, 50, 63],
    "lgb_model__max_depth": [-1, 5, 8, 12],
    "lgb_model__min_child_samples": [20, 30, 50, 70],
    "lgb_model__subsample": [0.6, 0.7, 0.8, 0.9, 1.0],
    "lgb_model__subsample_freq": [1],
    "lgb_model__colsample_bytree": [0.6, 0.7, 0.8, 0.9, 1.0],
    "lgb_model__reg_alpha": [0, 0.1, 0.5, 1.0],
    "lgb_model__reg_lambda": [0, 0.1, 0.5, 1.0, 2.0],
    "lgb_model__min_split_gain": [0, 0.01, 0.1]
}

# 6. Cross-validation and randomized search
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

random_search = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_distributions,
    n_iter=30,
    scoring="roc_auc",
    cv=cv,
    refit=True,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)

random_search.fit(X_fit, y_fit)

print("Best Parameters:")
print(random_search.best_params_)

print("\nBest Mean CV ROC-AUC:")
print(round(random_search.best_score_, 4))

# 7. Select the best model
best_model = random_search.best_estimator_

# 8. Choose classification threshold using validation data
y_prob_val = best_model.predict_proba(X_val)[:, 1]

threshold_results = []

for threshold in np.arange(0.10, 0.71, 0.01):
    y_pred_val = (
        y_prob_val >= threshold
    ).astype(int)

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Precision": precision_score(
            y_val, y_pred_val, zero_division=0
        ),
        "Recall": recall_score(
            y_val, y_pred_val, zero_division=0
        ),
        "F1": f1_score(
            y_val, y_pred_val, zero_division=0
        )
    })

threshold_results = pd.DataFrame(threshold_results)

best_row = threshold_results.loc[
    threshold_results["F1"].idxmax()
]

best_threshold = float(best_row["Threshold"])

print("\nBest Validation Threshold:", best_threshold)
print(best_row)

print("\nTop 10 Thresholds:")
print(
    threshold_results.sort_values(
        "F1", ascending=False
    ).head(10).round(4)
)

# 9. Evaluate once on untouched test data
y_prob_test = best_model.predict_proba(X_test)[:, 1]

y_pred_test = (
    y_prob_test >= best_threshold
).astype(int)

print("\nFinal LightGBM Test Results")
print("Threshold:", best_threshold)
print("Accuracy:", round(
    accuracy_score(y_test, y_pred_test), 4
))
print("ROC-AUC:", round(
    roc_auc_score(y_test, y_prob_test), 4
))
print("Precision:", round(
    precision_score(y_test, y_pred_test, zero_division=0), 4
))
print("Recall:", round(
    recall_score(y_test, y_pred_test, zero_division=0), 4
))
print("F1-score:", round(
    f1_score(y_test, y_pred_test, zero_division=0), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_test, zero_division=0
))

# 10. Confusion matrix
cm = confusion_matrix(y_test, y_pred_test)

sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("LightGBM - Test Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# 11. ROC curve
fpr, tpr, _ = roc_curve(y_test, y_prob_test)

plt.plot(
    fpr, tpr,
    label=f"ROC-AUC = {roc_auc_score(y_test, y_prob_test):.3f}"
)
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("LightGBM ROC Curve")
plt.legend()
plt.show()

# 12. Precision-recall curve
precision, recall, _ = precision_recall_curve(
    y_test, y_prob_test
)

plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("LightGBM Precision-Recall Curve")
plt.show()

# 13. Feature importance
model = best_model.named_steps["lgb_model"]
feature_names = best_model.named_steps[
    "processor"
].get_feature_names_out()

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Split_Importance": model.booster_.feature_importance(
        importance_type="split"
    ),
    "Gain_Importance": model.booster_.feature_importance(
        importance_type="gain"
    )
}).sort_values(
    "Gain_Importance", ascending=False
)

print("\nTop 15 Features:")
print(feature_importance.head(15))

lgb.plot_importance(
    model,
    importance_type="gain",
    max_num_features=15,
    figsize=(10, 8)
)
plt.title("LightGBM Feature Importance - Gain")
plt.show()
```
